```python
import numpy as np
```

```python
pt = np.array([10., 20., 30.])
pass_jet_id = np.array([True, True, False])

print("pt:", pt)
print("pass_jet_id:", pass_jet_id)
```

```python
mask_1 = pt > 15 & pass_jet_id
print("mask_1 =", mask_1)
```

```python
mask_2 = (pt > 15) & pass_jet_id
print("mask_2 =", mask_2)
```
